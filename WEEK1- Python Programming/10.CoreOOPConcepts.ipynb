```python
class Car:
    def __init__(self, make, model, year, colour):
        self.make = make
        self.model = model
        self.year = year
        self.colour = colour

    # def __init__(self,colour):
    #     self.colour = colour

    ## on runtime in pythoon the second method succeds the first one so the 
    ## second method will be used and the first one will be ignored.

    def display_info(self):
        return f"{self.year} {self.make} {self.model}"

    def CarColour(self):

        return f"The {self.display_info()} is {self.colour}."

    def OnlyCarColour(self):
    
            return f"The car is {self.colour}."
```

```python
car1 = Car("Toyota", "Camry", 2020, "Red")

car1.CarColour()
```
