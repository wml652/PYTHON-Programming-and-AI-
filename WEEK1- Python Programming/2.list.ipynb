```python
############## LIST #################

## ordered and changeable 

fruit = ["apple", "mango", "banana", "grapes","banana"]
print(fruit)
```

```python
#acessing elements 

print(fruit[0]) # first element 
print(fruit[2]) #third element  

print(fruit[-1]) # last element
```

```python
#methods to apply on lists

```

```python

#adding elements to list
fruit.append("orange")
print(fruit)
```

```python
#inserting at index 

fruit.insert(3, "guava")

print(fruit)
```

```python
#removing element

fruit.remove("orange") # removes first occurence of banana
print(fruit)
```

```python
helper = fruit.pop()
print(helper)

print(fruit)
```

```python
fruit.sort()
print(fruit)
```

```python
for i in fruit:
    print(i)
```
