```python
%pip install matplotlib
%pip install pandas
import pandas as pd
import matplotlib.pyplot as plt

# Sample data
days = [1, 2, 3, 4, 5, 6, 7]
sales = [120, 150, 180, 170, 200, 220, 210]
```

```python

plt.plot(days, sales,
marker='o',
linewidth=1.5,
alpha=0.8,
label="Daily Sales")

plt.xlabel("Day")
plt.ylabel("Sales")
plt.title("Sales Trend Over a Week")
plt.legend()
plt.show()
```

```python


# Sample data
temperature = [20, 22, 25, 27, 30, 32, 35]
ice_cream_sales = [50, 65, 80, 90, 120, 140, 160]

plt.scatter(temperature, ice_cream_sales,
alpha=0.7,
marker='x',
label="Sales vs Temp")

plt.xlabel("Temperature (℃)")
plt.ylabel("Ice Cream Sales")
plt.title("Relationship Between Temp & Sales")
plt.legend()
plt.show()

```

```python
df = pd.read_csv(
"ecommerce_sales_analytics_5000.csv",
parse_dates=["order_date"])

df = df.sample(n=200) 

# Line Plot - Monthly Revenue Trend
monthly_revenue = (
df
.set_index("order_date")["revenue"]
.resample("ME")
.sum()

)
```

```python
fig, ax = plt.subplots(figsize=(7, 4))
ax. plot(monthly_revenue. index, monthly_revenue. values, linewidth=2)

ax. set_title("Monthly Revenue Trend")
ax. set_xlabel("Month")
ax. set_ylabel("Total Revenue")

# Explanation: Line connects revenue over time and Shows growth or decline clearly
```

```python
# Scatter Plot - Discount vs Revenue
fig, ax = plt.subplots(figsize=(6, 4))
ax. scatter(df["unit_price"], df["revenue"], alpha=0.5)

ax. set_title("Unit Price vs Revenue")
ax. set_xlabel("Unit Price")
ax. set_ylabel("Revenue")

# Each point = one order and Transparency (alpha) helps with overlapping points
```
