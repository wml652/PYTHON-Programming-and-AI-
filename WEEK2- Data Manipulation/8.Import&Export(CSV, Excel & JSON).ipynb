```python
df = pd.read_csv("retail_sales.csv",parse_dates=["Date"],
dtype={
    "Category" : "category",
    "Region" : "category"
})

df.info()
```

```python
clean_df =  df[df["Sales"]>0]
subset = clean_df[["Date","Sales","Category","Region"]]
```

```python
subset.to_excel("clean_retail_sales.xlsx",sheet_name = 'sales')

subset.to_json("clean_retail_sales.json",
orient="records",
date_format="iso")

print("Files exported!")
subset.info()

```
