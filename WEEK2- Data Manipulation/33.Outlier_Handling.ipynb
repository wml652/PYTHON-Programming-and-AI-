```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
"sales": [100, 200, 150, 300, 5000, 700, 3000, 1500, 400, 600] # 5000 is an outlier
})

print(df.describe())
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set(style="whitegrid") 

plt.figure(figsize=(6,4))
sns.boxplot(y=df["sales"])
plt.title("Boxplot of Sales (Outliers visible)")
plt.show()
```

```python
# clipping
df['sales_clipped'] = df['sales'].clip(upper=1000)

print(df)
```

```python

import seaborn as sns
import matplotlib.pyplot as plt

sns.set(style="whitegrid")

plt.figure(figsize=(6,4))
sns.boxplot(data=df[[ "sales_clipped"]])
plt.title("Boxplot: original vs clipped sales (Upptr-1090)")
plt.show()
```

```python
from scipy.stats.mstats import winsorize

# Winsorization (top 20% capped)
df['sales_winsor'] = winsorize(df['sales'], limits=[0, 0.2])

print(df)
```

```python
# Removal (drop outliers > 1000)
df['sales_removed'] = df['sales'].where(df['sales'] <= 1000, np.nan)
print(df)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set(style="whitegrid")

plt.figure(figsize=(8,5))
sns.boxplot(data=df[["sales", "sales_clipped", "sales_winsor", "sales_removed"]])
plt.title("Boxplots: Original vs Clipped vs winsorized Sales vs Sales removed")
plt.show()

```
