```python
import numpy as np 
```

```python
# Simulate 100k coin flips
rng = np.random.default_rng()
flips = rng.integers(0, 2, size=100000) # 0 = tails, 1 = heads
prob_heads = np.mean(flips)
print(flips)
print("Estimated probability of heads:", prob_heads)
```

```python
# Simulate 100k coin flips with same randomness
rng = np.random.default_rng(seed = 50)
flips = rng.integers(0, 2, size=100000) # 0 = tails, 1 = heads
prob_heads = np.mean(flips)
print(flips)
print("Estimated probability of heads:", prob_heads)
```

```python
import time

# Python Loop
start = time.time()
squares = [i ** 2 for i in range(1000000)]
end = time.time()
print("Loop time:", end - start)

# NumPy vectorized
start = time.time()
arr = np.arange(1000000)
squares_np = arr ** 2
end = time.time()
print("Vectorized time:", end - start)


# as we can see that the numpy arrays are faster dur to vectoriation
```
